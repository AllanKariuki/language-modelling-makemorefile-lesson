```python
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
words = open('names.txt', 'r').read().splitlines()
```

```python
words[:8]
```

```python
# buld the vocabulary of characters and mappings to/fro integers
chars = sorted(list(set(''.join(words))))
stoi = {s:i+1 for i,s in enumerate(chars)}
stoi['.'] = 0
itos = {i:s for s,i in stoi.items()}
```

```python
# build the training dataset
block_size = 3 # the context length: how many characters do we take to predict the next one?
X, Y = [], [] # the input and the labels array, it's the actual dataset we're building
for w in words:
    # print(w)
    context = [0] * block_size # Show that we're at the beginning of a word [0, 0, 0]
    for ch in w + '.': # loop through the word and add a dot at the end of the word
        ix = stoi[ch]
        X.append(context)
        Y.append(ix)
        # print(''.join(itos[i] for i in context), '---->', itos[ix])
        context = context[1:] + [ix] # crop the first character then append ix
X = torch.tensor(X)
Y = torch.tensor(Y)
```

```python
# split the dataset into 80% training, 10% dev, 10% test
def build_dataset(words):
    block_size = 3 # the context length: how many characters do we take to predict the next one?
    X, Y = [], [] # the input and the labels array, it's the actual dataset we're building
    for w in words:
        # print(w)
        context = [0] * block_size # Show that we're at the beginning of a word [0, 0, 0]
        for ch in w + '.': # loop through the word and add a dot at the end of the word
            ix = stoi[ch]
            X.append(context)
            Y.append(ix)
            # print(''.join(itos[i] for i in context), '---->', itos[ix])
            context = context[1:] + [ix] # crop the first character then append ix
    X = torch.tensor(X)
    Y = torch.tensor(Y)
    print(X.shape, Y.shape)
    return X, Y

import random
random.seed(42)
random.shuffle(words)
n1 = int(0.8*len(words))
n2 = int(0.9*len(words))

Xtr, Ytr = build_dataset(words[:n1])
Xdev, Ydev = build_dataset(words[n1:n2])
Xte, Yte = build_dataset(words[n2:])
```

```python
g = torch.Generator().manual_seed(2147483647)
C = torch.randn((27,10), generator=g)
w1 = torch.randn((30, 200), generator=g)
b1 = torch.randn(200, generator=g)
w2 = torch.randn((200, 27), generator=g)
b2 = torch.randn(27, generator=g)
parameters = [C, w1, b1, w2, b2]
```

```python
sum(p.nelement() for p in parameters) # number of parameters in total
```

```python
for p in parameters:
    p.requires_grad= True
```

```python
lre = torch.linspace(-3, 0, 1000)
lrs = 10**lre
```

```python
lri = []
lossi = []
stepi = []
```

```python
for i in range(200000):

    # split the code into batches
    ix = torch.randint(0, Xtr.shape[0], (32,))
    # forward pass
    emb = C[Xtr[ix]] # (32, 3, 2)
    h = torch.tanh(emb.view(-1, 30) @ w1 + b1) # (32, 100)
    logits = h @ w2 + b2 # (32, 27)
    loss = F.cross_entropy(logits, Ytr[ix]) # new way of calculating loss
    
    # backward pass
    for p in parameters:
        p.grad = None
    loss.backward()

    # update
    # lr = lrs[i]
    lr = 0.1 if i < 100000 else 0.01
    for p in parameters:
        p.data += -lr * p.grad

    # track stats
    # lri.append(lre[i])
    stepi.append(i)
    lossi.append(loss.log10().item())
print(loss.item())
```

```python
plt.plot(stepi, lossi)
```
